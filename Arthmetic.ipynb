```python
# Define two sample numbers to work with
x = 10
y = 3

# 1. Addition (+)
addition = x + y
print(f"Addition: {x} + {y} = {addition}")

# 2. Subtraction (-)
subtraction = x - y
print(f"Subtraction: {x} - {y} = {subtraction}")

# 3. Multiplication (*)
multiplication = x * y
print(f"Multiplication: {x} * {y} = {multiplication}")

# 4. Division (/) - Always returns a decimal (float)
division = x / y
print(f"Division: {x} / {y} = {division}")

# 5. Floor Division (//) - Divides and rounds down to the nearest whole number
floor_division = x // y
print(f"Floor Division: {x} // {y} = {floor_division}")

# 6. Modulus (%) - Returns the leftover remainder of a division
remainder = x % y
print(f"Modulus (Remainder): {x} % {y} = {remainder}")

# 7. Exponentiation (**) - Raises a number to the power of another
power = x ** y
print(f"Exponent (Power): {x} ** {y} = {power}")
```
